### CSV DATA Downsampling (8,192)

##### Real World 1 (WOODY_DOWNSAMPLED)

```python
import os
import pandas as pd
import numpy as np
from glob import glob
import torch
import sys
sys.path.append('/esail4/heeju/Point-M2AE/')
from utils.misc import fps

source_dir = "/bess25/heeju/DATA/LABELX/WOODY_DOWNSAMPLED"
target_dir = "/bess25/heeju/DATA/REGRESSION/M2AE_PRETRAIN/PROCESSED/DOWNSAMPLED/RW01_8192_LX"

sample_size = 8192

os.makedirs(target_dir, exist_ok=True)

csv_files = sorted(glob(os.path.join(source_dir, "*.csv")))

for i, file in enumerate(csv_files):
    print(f'{file}')
    df = pd.read_csv(file)
    lw_df = df[["X", "Y", "Z"]]
    lw_points = torch.tensor(lw_df.values).cuda().float().contiguous().unsqueeze(0)

    if len(lw_points[0]) >= sample_size:
        sampled_points = fps(lw_points, sample_size)
        sampled_points = sampled_points.cpu().numpy().squeeze()
        sampled_df = pd.DataFrame(sampled_points, columns=["X", "Y", "Z"])
    else:
        sampled_df = pd.DataFrame(lw_points.cpu().numpy().squeeze(), columns=["X", "Y", "Z"])  # 안전상 처리 (매우 적은 경우)
        
    print(sampled_df.shape)
    filename = f"{str((i + 1) * 10000).zfill(8)}.csv"
    save_path = os.path.join(target_dir, filename)
    sampled_df.to_csv(save_path, index=False)
```

#### train test split

```python
import os
import glob
import numpy as np
import math

# Input folder containing CSV files
tree_csv_path = 'dataset/tls_simul'

# Collect all CSV files
tree_files = glob.glob(os.path.join(tree_csv_path, '*.csv'))
tree_names = [os.path.basename(file_path) for file_path in tree_files]

total_files = len(tree_names)

# Group IDs (here simply using file names as IDs)
group_ids = [name for name in tree_names]

# Unique group IDs
unq_group_ids = np.unique(group_ids)
id_length = len(unq_group_ids)

# Split ratios
train_ratio, test_ratio, extra_ratio = 0.7, 0.15, 0.15

# Number of files per split
train_count = math.ceil(id_length * train_ratio)
test_count = math.ceil(id_length * test_ratio)
extra_count = math.ceil(id_length * extra_ratio)

# NOTE: Shuffle for random split if needed
import random
random.shuffle(unq_group_ids)

# Split into train / test / extra
test_set = unq_group_ids[:test_count]
train_set = unq_group_ids[test_count:-extra_count]
extra_set = unq_group_ids[-extra_count:]

# Helper: expand IDs back to filenames
def expand_ids_to_files(id_set, group_ids, tree_names):
    expanded = []
    for ids in id_set:
        indices = np.where(np.array(group_ids) == ids)[0]
        expanded += [tree_names[i] for i in indices]
    return expanded

train_set_list = expand_ids_to_files(train_set, group_ids, tree_names)
test_set_list = expand_ids_to_files(test_set, group_ids, tree_names)
extra_set_list = expand_ids_to_files(extra_set, group_ids, tree_names)

print(f"Train groups: {len(train_set)}")
print(f"Test groups: {len(test_set)}")
print(f"Extra groups: {len(extra_set)}")

# Destination folder and text files
destination = 'splits/tls_simul_txt'
os.makedirs(destination, exist_ok=True)

train_txt_path = os.path.join(destination, 'train.txt')
test_txt_path = os.path.join(destination, 'test.txt')
extra_txt_path = os.path.join(destination, 'extra_train.txt')

# Save splits
with open(train_txt_path, 'w') as f:
    f.write('\n'.join(train_set_list))

with open(test_txt_path, 'w') as f:
    f.write('\n'.join(test_set_list))

with open(extra_txt_path, 'w') as f:
    f.write('\n'.join(extra_set_list))

print("Split files saved to:", destination)
```
